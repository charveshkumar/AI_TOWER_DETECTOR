### Step 2: Upload and Unzip Dataset (`yolo_dataset.zip`)
Run this cell and upload the `yolo_dataset.zip` file generated in your local project folder.

```python
from google.colab import files
import zipfile, os

if not os.path.exists('yolo_dataset.zip'):
    print('Please upload yolo_dataset.zip:')
    uploaded = files.upload()

# Extract dataset
!mkdir -p /content/dataset
with zipfile.ZipFile('/content/yolo_dataset.zip', 'r') as zip_ref:
    zip_ref.extractall('/content/dataset')

print('Dataset extracted successfully!')
!ls -la /content/dataset
```

### Step 3: Configure `data.yaml` with Absolute Colab Paths

```python
yaml_content = """
path: /content/dataset
train: train/images
val: val/images
test: test/images
nc: 2
names:
  - supporting_tower
  - monopole_tower
"""

with open('/content/dataset/data.yaml', 'w') as f:
    f.write(yaml_content.strip())

print('data.yaml configured:')
!cat /content/dataset/data.yaml
```

### Step 4: Train YOLO11m Model on GPU (~2-3 minutes total)

```python
from ultralytics import YOLO

# Load pretrained YOLO11m
model = YOLO('yolo11m.pt')

# Train on GPU
results = model.train(
    data='/content/dataset/data.yaml',
    epochs=100,
    imgsz=640,
    batch=16,
    patience=20,
    device=0,      # GPU device
    seed=42,
    project='/content/runs/detect',
    name='yolo11m_baseline',
    save=True,
    plots=True
)

print('Training complete!')
```

### Step 5: Download Trained Checkpoints and Results to Local Machine

```python
import shutil
from google.colab import files

# Zip the output directory
shutil.make_archive('/content/yolo11m_baseline_results', 'zip', '/content/runs/detect/yolo11m_baseline')

# Download the zip
files.download('/content/yolo11m_baseline_results.zip')
```
